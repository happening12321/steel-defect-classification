```python
import pandas as pd
from ucimlrepo import fetch_ucirepo

# 데이터 불러오기
data = fetch_ucirepo(id=198)

# 독립변수와 종속변수
X = data.data.features
y = data.data.targets

# 하나의 데이터프레임으로 합치기
df = pd.concat([X, y], axis=1)

# 데이터 확인
display(df.head())
```

```python
# 데이터 크기 확인
print(df.shape)

# 데이터 타입 확인
df.info()

# 상위 5개 행 확인
display(df.head())

# 변수명 확인
print(df.columns.tolist())
```

```python
# 결측치 확인
print(df.isnull().sum())

# 전체 결측치 개수
print("전체 결측치:", df.isnull().sum().sum())

# 중복 데이터 개수
print("중복 데이터:", df.duplicated().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 독립변수와 종속변수 분리
X = df.iloc[:, :27]
y = df.iloc[:, 27:]

# 불량 유형 하나의 열로 변환
fault_type = y.idxmax(axis=1)

# 불량 유형별 개수
print(fault_type.value_counts())

# 불량 유형별 비율
print(fault_type.value_counts(normalize=True) * 100)

# 시각화
plt.figure(figsize=(10, 5))

sns.countplot(
    x=fault_type,
    order=fault_type.value_counts().index
)

plt.xticks(rotation=45)
plt.title("Distribution of Fault Types")
plt.xlabel("Fault Type")
plt.ylabel("Count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 독립변수만 선택
X = df.iloc[:, :27]

# 변수별 박스플롯
fig, axes = plt.subplots(7, 4, figsize=(16, 24))

axes = axes.flatten()

for i, col in enumerate(X.columns):
    sns.boxplot(y=X[col], ax=axes[i])
    axes[i].set_title(col)

# 남는 그래프 칸 제거
for j in range(len(X.columns), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# IQR 기준 이상치 개수 계산
for col in X.columns:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = X[(X[col] < lower) | (X[col] > upper)]

    print(f"{col}: {len(outliers)}개")
```

```python
plt.figure(figsize=(14, 10))

corr = X.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")
plt.show()
```
