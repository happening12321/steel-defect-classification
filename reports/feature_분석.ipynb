```python
# X_Minimum              결함 bounding box의 가장 왼쪽 X 좌표
# X_Maximum              결함 bounding box의 가장 오른쪽 X 좌표
# Y_Minimum              결함 bounding box의 Y 시작 좌표 (추정: 강판 진행 방향 누적 좌표)
# Y_Maximum              결함 bounding box의 Y 끝 좌표
# Pixels_Areas           결함이 차지하는 픽셀 수 (면적)
# X_Perimeter            (추정) X 방향 윤곽 길이. 항상 ≥ Δx, 윤곽이 복잡할수록 커짐
# Y_Perimeter            (추정) Y 방향 윤곽 길이. 항상 ≥ Δy
# Sum_of_Luminosity      결함 영역 픽셀 밝기의 합
# Minimum_of_Luminosity  결함 영역의 최소 밝기 (0~255 회색조 추정)
# Maximum_of_Luminosity  결함 영역의 최대 밝기
# Length_of_Conveyer     (추정) 강판의 X 방향 전체 길이(폭). X_Maximum ≤ L
# TypeOfSteel_A300       강종 A300 여부 (원-핫, A400과 합이 항상 1)
# TypeOfSteel_A400       강종 A400 여부
# Steel_Plate_Thickness  강판 두께
# Edges_Index            2·min(X_Min, L − X_Max) / L : 강판 가장자리와의 거리 (0=가장자리, 1=중앙)
# Empty_Index            1 − 면적/(Δx·Δy) : bounding box 중 빈 공간 비율
# Square_Index           min(Δx,Δy)/max(Δx,Δy) : 정사각형에 가까운 정도 (1=정사각형)
# Outside_X_Index        Δx / L : 결함 폭이 강판 폭에서 차지하는 비율
# Edges_X_Index          Δx / X_Perimeter : (추정) X 방향 윤곽의 매끈함 (1=곧음)
# Edges_Y_Index          Δy / Y_Perimeter : (추정) Y 방향 윤곽의 매끈함 (1=곧음)
# Outside_Global_Index   Δx>Δy면 0, Δx=Δy면 0.5, Δy>Δx면 1 : 결함이 뻗은 주 방향 (Orientation_Index의 부호)
# LogOfAreas             log10(Pixels_Areas) : 면적 로그 변환 (왜도 14.1 → 0.75)
# Log_X_Index            log10(Δx) : 가로 길이 로그 변환
# Log_Y_Index            log10(Δy) : 세로 길이 로그 변환
# Orientation_Index      Δy≥Δx면 1−Δx/Δy, 아니면 Δy/Δx−1 : 방향성 (+:Y로 길쭉, −:X로 길쭉)
# Luminosity_Index       (평균 밝기 − 128)/128 : 평균 밝기를 −1~1로 정규화 (음수=어두움)
# SigmoidOfAreas         (추정) 1/(1+exp(−(Δx·Δy − 302.25)/150)) : bounding box 넓이의 시그모이드 변환 (중심≈박스 넓이 중앙값, 큰 결함은 1로 포화)
# ※ Δx = X_Maximum − X_Minimum, Δy = Y_Maximum − Y_Minimum, L = Length_of_Conveyer
# ※ 계산식은 공식 정의가 아니라 데이터 1,941행 전체에 대입해 일치를 확인한 역산 결과
```
